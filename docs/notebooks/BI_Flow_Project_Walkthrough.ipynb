# BI Flow â€” Dashboard Generator Agent: full project walkthrough

This notebook explains and runs, step by step, everything the **Dashboard Generator Agent** does:
how it reads the upstream outputs, computes KPIs and DAX, retrieves domain knowledge (RAG), uses an
LLM layer, plans and scores several dashboard designs, validates them, and finally builds the
**Power BI project (PBIP)** and the **offline HTML dashboard**.

Every cell runs offline on the files in `input/`. No API key is needed: the LLM layer uses a deterministic
mock unless you set `LLM_API_KEY` (see section 4).

## Where this agent sits in BI Flow

BI Flow automates the Power BI pipeline with one specialised agent per Power BI step, coordinated by an orchestrator:

| # | Agent | Replaces in Power BI | Output consumed here |
|---|-------|----------------------|----------------------|
| 1 | Data Profiler | Get Data | â€” |
| 2 | Quality & Cleaning | Power Query | `cleaned_trade.csv`, `data_quality_report.json` |
| 3 | Semantic & KPI | Data model + DAX | `data_model.json`, `kpi_catalog.json` |
| 4 | BI Analyst | Insights | `bi_analyst_findings.json` |
| **5** | **Dashboard Generator (this project)** | **Visuals (PBIP)** | â€” |
| 6 | BI Auditor / XAI | Traceability | `xai_lineage.json` |
| 0 | Orchestrator (LangGraph) | Coordination | runs the agents in order |

### What happens with bad data?

`validate_frame` normalises column names, parses dates and numbers, checks the `Import`/`Export` domain
and reports every problem with its rows. Errors stop the generation (`VALIDATION_FAILED`);
warnings are kept and reported.

````python
bad = pd.DataFrame({
    "date": ["2024-01-01", "not a date", "2024-03-01"],
    "country": ["DE", "US", "FR"],
    "product": ["Machinery", "Chemicals", "Electronics"],
    "direction": ["Import", "Sideways", "Export"],
    "value": [100, "abc", -5],
})
from src.agents.dashboard.dashboard_generator import InputValidationError
try:
    validate_frame(bad)
except InputValidationError as exc:
    print("Rejected:", exc)
    display(pd.DataFrame([i.as_dict() for i in exc.issues])[["code", "severity", "column", "rows", "message"]])
````

## 3. RAG: retrieving domain knowledge

`KnowledgeAgent` indexes the approved documents in `resources/knowledge/`. Each markdown section becomes a
chunk. A query is scored against every chunk with a TF-IDF formula (term frequency Ã— inverse document
frequency), with a Ã—2 boost for words found in the section title. The best chunks are given to the planner,
which writes them into each visual's evidence as *domain guidance*.

> Current limit: this is lexical retrieval. The project report targets a vector store (e.g. ChromaDB) with
> embeddings; the `retrieve()` interface stays the same, so it can be swapped later.

````python
from knowledge_agent import KnowledgeAgent
from src.agents.dashboard.config import GeneratorConfig

cfg = GeneratorConfig()
kb = KnowledgeAgent(cfg.knowledge_dir)
print(f"{len(kb.chunks)} chunks indexed from {len(set(c.doc_id for c in kb.chunks))} documents")
display(pd.DataFrame([(c.doc_id, c.chunk_id, c.title, len(c.content)) for c in kb.chunks],
                     columns=["document", "chunk", "section title", "chars"]))
````

````python
query = "compare partner countries and export concentration"
hits = kb.retrieve(query, top_k=3)
print("Query:", query)
for h in hits:
    print(f"\n--- {h.source} / {h.chunk.title}  (score {h.score:.3f})")
    print(h.chunk.content[:400])
````

## 4. The LLM layer and the Dashboard Copilot

All LLM calls go through one interface, `LLMProvider.generate(prompt, system_prompt) -> str`:

| Provider | When | Behaviour |
|----------|------|-----------|
| `MockLLMProvider` | `llm_provider="mock"` (default) | Offline, deterministic, used by tests and demos |
| `HTTPLLMProvider` | `llm_provider="openai"` or `"custom"` | OpenAI-compatible `/chat/completions`, JSON mode, timeout |

The **Copilot** turns the user's request into a `DashboardSpecification` (objective, audience, questions,
required KPIs, preferred pages and visuals). Its guardrails:

* only KPI ids from the catalog are accepted; anything else goes to `unresolved_questions`;
* if the LLM answer is not a specification (e.g. the mock), it falls back to rule-based extraction of the request;
* if the LLM fails (bad JSON, network, missing key), the planner falls back to the deterministic plan.

````python
from src.agents.dashboard.dashboard_copilot import DashboardCopilot
from src.agents.dashboard.dashboard_planner import MockLLMProvider, HTTPLLMProvider

request = "Executive dashboard for export managers comparing partner countries and analyst findings"
spec = DashboardCopilot(MockLLMProvider()).interpret(request, bundle)
print(json.dumps(spec.model_dump(), indent=2))
````

````python
# Requests for metrics the upstream catalog does not have are flagged, never invented
spec_unsupported = DashboardCopilot(MockLLMProvider()).interpret("Show profit margin and revenue forecast by country", bundle)
print("Required KPIs:", spec_unsupported.required_kpi_ids)
print("Unresolved:", *spec_unsupported.unresolved_questions, sep="\n - ")
````

## 5. Multi-candidate planning

From the specification and the RAG chunks, the planner builds three deliberately different designs:

* **Executive** â€” headline KPI cards, macro trend, governance page;
* **Analytical** â€” country and product breakdowns with detail tables;
* **Decision** â€” analyst findings first, bilateral risk drill-downs.

Each visual carries a `VisualEvidence` record: the business question, the KPI or insight, why this chart type,
the upstream source, the domain rule from RAG, and an evidence category
(`UPSTREAM_FACT`, `DOMAIN_GUIDANCE`, `AI_INTERPRETATION`, `RECOMMENDATION`, `UNKNOWN`).

````python
from src.agents.dashboard.dashboard_planner import generate_candidate_plans

candidates = generate_candidate_plans(bundle, spec, hits, MockLLMProvider(), candidate_count=3)
rows = []
for c in candidates:
    for p in c.pages:
        rows.append((c.metadata.get("candidate_id"), p.display_name, ", ".join(f"{v.id}:{v.visual_type}" for v in p.visuals)))
display(pd.DataFrame(rows, columns=["candidate", "page", "visuals (id:type)"]))
````

````python
# Evidence attached to one visual
v = next(v for c in candidates for p in c.pages for v in p.visuals if v.evidence)
print(v.id, "-", v.title)
print(json.dumps(v.evidence.model_dump(mode="json"), indent=2))
````

## 6. Scoring the candidates

`DashboardEvaluator` scores each candidate from 0 to 100 on 10 weighted criteria (weights from `config.py`,
renormalised to sum to 1). A candidate that fails a hard rule (unknown table, column or measure, duplicate
visual id) is disqualified whatever its score. The highest eligible score is selected.

````python
from src.agents.dashboard.dashboard_evaluator import DashboardEvaluator

evaluator = DashboardEvaluator(weights=cfg.evaluator_weights)
summary = evaluator.evaluate_candidates(candidates, bundle, spec)
display(pd.Series(evaluator.weights, name="normalised weight").round(3).to_frame())
matrix = pd.DataFrame(summary.comparison_matrix)
matrix.loc["TOTAL"] = [e.total_score for e in summary.evaluations]
display(matrix)
print("Selected:", summary.selected_candidate_id)
print(summary.selection_rationale)
````

## 7. The critic: hard validation and revision

`DashboardCritic` checks every reference of a plan against the data contract. Below it accepts the
selected plan, then rejects a copy where one visual points to a table that does not exist.

````python
from src.agents.dashboard.dashboard_critic import DashboardCritic

critic = DashboardCritic()
selected = next(c for c in candidates if c.metadata.get("candidate_id") == summary.selected_candidate_id)
report = critic.evaluate(selected, bundle)
print("Selected plan ->", report.status, "|", *report.quality_feedback)

broken = selected.model_copy(deep=True)
vis = next(v for p in broken.pages for v in p.visuals if v.roles)
role = next(iter(vis.roles))
vis.roles[role][0]["table"] = "FakeTable"
bad_report = critic.evaluate(broken, bundle)
print("Broken plan   ->", bad_report.status)
print(*bad_report.errors, sep="\n - ")
````

## 8. Optimizer and feedback log

The optimizer tries safe operations in priority order (remove redundant visuals, reorder pages for the
audience, add a missing KPI card, add missing evidence, move a misplaced insights visual). An operation is
kept only if the plan still passes the critic and the score does not drop. Every selection and
optimisation result is appended to a JSONL feedback log for later analysis.

````python
from src.agents.dashboard.dashboard_optimizer import DashboardOptimizer
from src.agents.dashboard.dashboard_feedback import FeedbackLogger

optimizer = DashboardOptimizer(evaluator=evaluator, critic=critic, max_operations=5)
optimized, opt_report = optimizer.optimize(candidates, selected, bundle, spec, summary, user_request=request)
print(f"baseline {opt_report.baseline_score} -> optimized {opt_report.optimized_score} (improved={opt_report.improved})")
for op in opt_report.operations:
    print(f" - {op.operation_type}: applied={op.applied} {op.rejection_reason or ''}")
print("Findings:", [f.defect for f in opt_report.findings] or "none")

log_path = Path(tempfile.mkdtemp()) / "feedback.jsonl"
fb = FeedbackLogger(log_path)
fb.log_candidate_selected(run_id=opt_report.run_id, candidate_id=summary.selected_candidate_id,
                          strategy_name=summary.selected_strategy, score=opt_report.optimized_score,
                          reason=summary.selection_rationale)
print("Feedback events:", fb.count_events_by_type())
````

## 9. The whole planning loop in one call, and its safety net

`DashboardPlanner.plan()` chains sections 4 to 8. In deterministic mode it skips the LLM and returns the
standard verified 6-page plan. If anything fails in AI mode, it returns that deterministic plan with
`planning_mode = "deterministic_fallback"` and the reason in its warnings.

````python
from src.agents.dashboard.config import OperatingMode
from src.agents.dashboard.dashboard_planner import DashboardPlanner

def ai_cfg(**kw):
    return GeneratorConfig(mode=OperatingMode.AI_ASSISTED, llm_provider="mock",
                           feedback_path=Path(tempfile.mkdtemp()) / "fb.jsonl", **kw)

rows = []
for label, planner, req in [
    ("deterministic", DashboardPlanner(GeneratorConfig()), request),
    ("AI, countries", DashboardPlanner(ai_cfg()), request),
    ("AI, products", DashboardPlanner(ai_cfg()), "Which products and commodities drive exports?"),
    ("AI, LLM returns bad JSON", DashboardPlanner(ai_cfg(), provider=MockLLMProvider(response_override="not json")), request),
    ("AI, LLM network error", DashboardPlanner(ai_cfg(), provider=MockLLMProvider(raise_error=TimeoutError("down"))), request),
]:
    plan = planner.plan(bundle, req, hits)
    rows.append((label, plan.planning_mode, len(plan.pages), ", ".join(p.display_name for p in plan.pages)))
display(pd.DataFrame(rows, columns=["scenario", "planning_mode", "pages", "page names"]))
````

## 12. End to end with `main()` in every mode

`main()` is what the orchestrator (or a user) calls: it loads the inputs, plans in the configured mode,
generates the PBIP and HTML, and runs the checks. It never raises on bad input: it returns a
`VALIDATION_FAILED` report instead.

````python
import contextlib, io, shutil
from src.agents.dashboard.dashboard_generator import main

def run(label, **kw):
    out = Path(tempfile.mkdtemp())
    if isinstance(kw.get("config"), GeneratorConfig):
        kw["config"].feedback_path = out / "fb.jsonl"
    with contextlib.redirect_stdout(io.StringIO()):
        r = main(output_root=out, **kw)
    return (label, r["status"], r.get("planning_mode", "-"), len(r["report_pages"]), (r["errors"] or [{}])[0].get("message", ""))

broken_input = Path(tempfile.mkdtemp()) / "input"
shutil.copytree(INPUT, broken_input)
(broken_input / "kpi_catalog.json").unlink()

display(pd.DataFrame([
    run("deterministic", config=GeneratorConfig()),
    run("AI assisted", config=GeneratorConfig(mode=OperatingMode.AI_ASSISTED, llm_provider="mock"), user_context=request),
    run("AI without RAG", config=GeneratorConfig(mode=OperatingMode.AI_ASSISTED_WITHOUT_RAG, llm_provider="mock"), user_context=request),
    run("demo fixture", config=GeneratorConfig(), demo_mode=True),
    run("missing kpi_catalog.json", config=GeneratorConfig(), input_root=broken_input),
], columns=["run", "status", "planning_mode", "pages", "error"]))
````
